This notebook estimates the intention-to-treat effects of the men's and women's email treatments against the no-email control. The primary commercial outcome is spend per eligible customer; visit and conversion are supporting outcomes.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd
from src.data_validation import load_data, validate_data
from src.experiment_metrics import arm_summary, main_effect_table

DATA_PATH = PROJECT_ROOT / 'data/raw/hillstrom.csv'
frame = load_data(DATA_PATH)
validation = validate_data(frame)
validation.to_dict()
```

## Analysis population

## Interpretation checklist

For each treatment, interpret the point estimate, 95% confidence interval and commercial scale together. A small p-value alone is not sufficient for a campaign recommendation. Revenue per eligible customer is the primary outcome; visit and conversion explain the pathway but do not replace the commercial result.
